# P-value Calculator:

```python
from scipy import stats
```

```python
'''
x_mean - sample mean / новое среднее
M_mean(μ) - population mean / среднее в генеральной совокупности
sd - standard deviation / стандартное отклонение в выборке
n - sample size / количество выборочных значений
'''
```

```python
def calc_pvalue_norm(M_mean, x_mean, sd, n):
    se = sd/(n**(1/2))
    z = abs((x_mean-M_mean)/se)
    p = (1 - stats.norm.cdf(z))*2
    return p
```

```python
print('p = {:.5f}'.format(calc_pvalue_norm(102, 98, 10, 64)))
```

```python
'''
On average, participants of the introductory statistics course score 115 points.
However, in 2015 the average score of a randomly selected group of 144 participants
was 118, with a standard deviation of 9.

Calculate the p-value to test the null hypothesis that the mean score
in 2015 is equal to 115.
'''

'''
В среднем слушатели курса по введению в статистику набирают 115 баллов,
однако, в 2015 году средний балл  случайно выбранных 144 участников
составил 118 со стандартным отклонением равным 9.

Рассчитайте p уровень значимости для проверки нулевой гипотезы о том,
что среднее значение баллов в 2015 году равняется 115.
'''
```

```python
print('p = {:.5f}'.format(calc_pvalue_norm(115, 118, 9, 144)))
```
